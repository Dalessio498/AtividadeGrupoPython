```python
import pandas as pd
# --- Amostra 1: Desempenho de Alunos ---
dados_alunos = {
'aluno': ['Ana', 'Bruno', 'Carla', 'David', 'Elisa', 'Fábio', 'Gabriela', 'Helio'],
'turma': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
'nota_final': [8.5, 7.0, 9.0, 5.5, 9.5, 6.5, 7.5, 8.0],
'faltas': [2, 5, 1, 6, 0, 4, 3, 2]
}
df_alunos = pd.DataFrame(dados_alunos)
print("--- Amostra 1: Desempenho de Alunos ---")
print(f"{df_alunos}\n")

listaNomes = ["Carla", "Emmilly", "Lucas"]

for nome in listaNomes:
    print(f"Boas vindas {nome}!\n")

for rank in range(1, 4):
    print(f"Rank {rank}")

print("\n--- Média de notas por turma ---")
mediaNotas = df_alunos.groupby('turma')['nota_final'].mean()
print(mediaNotas)
print("Turma A teve melhor desempenho")

print("\n--- Soma de faltas por turma ---")
somaNotas = df_alunos.groupby('turma')['faltas'].sum()
print(somaNotas)

print("\n--- Total de alunos por turma ---")
totalNotas = df_alunos.groupby('turma')['aluno'].count()
print(totalNotas)
```
