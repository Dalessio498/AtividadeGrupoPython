```python


# Dados para Pessoa A (Marketing)
dados_marketing = pd.DataFrame({
'investimento_ads_dia': [10, 15, 20, 22, 25, 30, 33, 35, 40, 41, 45, 50, 55, 60, 62],
'visitantes_unicos': [1100, 1200, 1500, 1600, 1550, 1800, 1900, 2000, 2200, 2150, 2400, 2800, 3100, 3500, 3400]
})
# Dados para Pessoa B (Logística)
dados_logistica = pd.DataFrame({
'transportadora_X': [3, 4, 4, 5, 5, 5, 5, 6, 6, 6, 7, 7, 8, 8, 9],
'transportadora_Y': [1, 2, 2, 3, 4, 5, 5, 6, 7, 8, 9, 10, 10, 11, 12]
})
MeuPaleta = ['#1f1f20','#2b4c7e', '#606d80', '#dce0e6']
(sns.set_theme(style="whitegrid"))
plt.figure(figsize=(10, 6))

sns.boxplot(
    palette = MeuPaleta,
    data = dados_logistica,
    x = 'transportadora_X',
    y = 'transportadora_Y',
    linewidth = 2.5
)

plt.title('Grafico transportadora')

plt.show()  # ← ESSA LINHA É OBRIGATÓRIA!


media_transportadoraA = dados_logistica ['transportadora_X'].mean()
print(f"A media da transportadora 1 é {media_transportadoraA:.2f} ")

desvio_transportadoraA = dados_logistica ['transportadora_X'].std()
print(f"O desvio padrão da transportadora 1 é {desvio_transportadoraA:.2f} \n")

media_transportadoraB = dados_logistica ['transportadora_Y'].mean()
print(f"A media da transportadora 2 é {media_transportadoraB:.2f} ")

desvio_transportadoraB = dados_logistica ['transportadora_Y'].std()
print(f"O desvio padrão da transportadora 2 é {desvio_transportadoraB:.2f} \n")

# è visivel que a transportadora 2 tem uma media maior de entregas, porem, a transportadora 1 tem uma maior constancia em entregas
```

```python

#Bloco 1

# Investimentos 

mediaInvestimento = dados_marketing['investimento_ads_dia'].mean()
print(f"Media Investimento: {mediaInvestimento:.2f} ")

medianaInvestimento = dados_marketing['investimento_ads_dia'].median()
print(f"Mediana Investimento: {medianaInvestimento:.2f} \n")

# ------------------------------------------------------------------------

# Visitantes

mediaVisitantes = dados_marketing['visitantes_unicos'].mean()
print(f"Media Visitante Criada: {mediaVisitantes:.2f}")

medianaVisitante = dados_marketing['visitantes_unicos'].median()
print(f"Mediana Visitante: {medianaVisitante:.2f}")

# ------------------------------------------------------------------------

sns.set_style('darkgrid')

plt.figure(figsize=(10, 6))

sns.regplot(
    data = dados_marketing,
    x = 'investimento_ads_dia',
    y = 'visitantes_unicos',

color="#ff7b00",

scatter_kws = {

        's' : 60,
        'color': "#f8ab47",
        'marker' : 'x',
        'edgecolor' : "#242424"
    } ,

    line_kws= {
        'color' : "#FF9100",
        'linewidth' : 4
    } 
)

plt.title('Gráfico marketing', fontsize = 16, fontweight = 'bold')
plt.xlabel('Investimento diário', fontsize = 16)
plt.ylabel('Visitantes unicos', fontsize = 16)
plt.grid(True, linestyle='--', alpha=1) # Deixa a grade tracejada e semitransparente



#Sim existe correlação, ela é boa pois compara duas variaves assim mostrando com o grafico qual das duas se sobrepõe
```
