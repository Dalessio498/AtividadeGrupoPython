```python
import pandas as pd

dados_pets = {
'NomePet': ['Floco', 'Rex', 'Mia', 'Thor', 'Bolinha', 'Luna'],
'Especie': ['Cachorro', 'Cachorro', 'Gato', 'Cachorro', 'Gato', 'Cachorro'],
'Idade': [3, 8, 2, 12, 5, 1],
'Peso_kg': [15, 25, 4, 30, 6, 8],
'Vacinado': [True, True, False, True, True, False]
}

df_pets = pd.DataFrame(dados_pets)

print("--- Clínica Veterinária ---")
print(f"{df_pets} \n")

pets_gatos = df_pets['Especie'] == 'Gato'
print(df_pets[pets_gatos])
print("\n")

pets_pesados = df_pets['Peso_kg'] > 20
print(df_pets[pets_pesados])

pets_grandes = df_pets['Especie'] == 'Cachorro'
pets_idosos = df_pets['Idade'] > 7

pets_idosos_grandes = df_pets[pets_idosos & pets_grandes]

print(pets_idosos_grandes)
print("\n")



```

```python
dados_estoque = {
'Produto': ['Arroz', 'Feijão', 'Óleo', 'Sal', 'Macarrão', 'Café'],
'Categoria': ['Grãos', 'Grãos', 'Mercearia', 'Mercearia', 'Massas', 'Mercearia'],
'Qtd_Estoque': [50, 15, 10, 100, 40, 5],
'Preco_Unit': [25.00, 8.00, 7.50, 3.00, 6.00, 15.00]
}
df_estoque = pd.DataFrame(dados_estoque)
print("\n--- Controle de Estoque ---")
print(df_estoque)
print("\n")

estoqueBaixo = df_estoque['Qtd_Estoque'] <= 10
print(df_estoque[estoqueBaixo])
print("\n")

marceariaEstoque = df_estoque['Categoria'] == 'Marcearia'
graosEstoque = df_estoque['Categoria'] == 'Grãos'

reposicaoUrgente = df_estoque[marceariaEstoque | graosEstoque]
print(reposicaoUrgente)

print("\n")

graosEstoque = df_estoque['Categoria'] == 'Grãos'
precoUnit = df_estoque['Preco_Unit'] < 10

ofertaGraos = df_estoque[graosEstoque & precoUnit]

print(ofertaGraos)
print("\n")


```

```python
dados_alunos = {
'Matricula': [101, 102, 103, 104, 105, 106],
'Nome': ['Ana', 'Bruno', 'Carla', 'Daniel', 'Elisa', 'Fábio'],
'Nota_Final': [8.5, 6.0, 9.5, 4.0, 7.0, 5.5],
'Faltas': [2, 5, 0, 10, 1, 3]
}
df_alunos = pd.DataFrame(dados_alunos)
print("\n--- Desempenho Acadêmico ---")
print(df_alunos)
print("\n")

alunosAprovados = df_alunos['Nota_Final'] >= 7
print(df_alunos[alunosAprovados])
print("\n")

alunosReprovadosFalta = df_alunos['Faltas'] > 5
print(df_alunos[alunosReprovadosFalta])
print("\n")

alunosExemplaresNota = df_alunos['Nota_Final'] > 9
alunosExemplaresFalta = df_alunos['Faltas'] == 0

alunoExemplar = df_alunos[alunosExemplaresNota & alunosExemplaresFalta]

print(alunoExemplar)
```

```python
dados_vendas = {
'Vendedor': ['Rafael', 'Bia', 'Carlos', 'Rafael', 'Bia', 'Carlos'],
'Regiao': ['Sul', 'Sudeste', 'Nordeste', 'Sudeste', 'Sul', 'Sudeste'],
'Valor_Venda': [1200, 1500, 800, 2000, 1800, 1300],
'Tipo': ['Online', 'Presencial', 'Online', 'Online', 'Presencial', 'Presencial']
}
df_vendas = pd.DataFrame(dados_vendas)
print("\n--- Vendas da Equipe ---")
print(df_vendas)
print("\n")

vendasBia = df_vendas['Vendedor'] == 'Bia'
print(df_vendas[vendasBia])
print("\n")

vendasAltas = df_vendas['Valor_Venda'] > 1400
print(df_vendas[vendasAltas])
print("\n")

regiaoSul = df_vendas['Regiao'] == 'Sul'
regiaoNorte = df_vendas['Regiao'] == 'Norte'

vendasFoco = df_vendas[regiaoSul | regiaoNorte]
print(vendasFoco)
print("\n")

vendasOffline = df_vendas['Tipo'] != 'Online'
print(df_vendas[vendasOffline])
print("\n")




```

```python
dados_imoveis = {
'Bairro': ['Centro', 'Praia', 'Centro', 'Norte', 'Praia', 'Sul'],
'Tipo': ['Apartamento', 'Casa', 'Apartamento', 'Apartamento', 'Casa', 'Casa'],
'Area_m2': [60, 120, 45, 80, 150, 90],
'Preco_Aluguel': [1500, 2800, 1200, 1600, 3500, 2000],
'Quartos': [2, 3, 1, 2, 4, 3]
}
df_imoveis = pd.DataFrame(dados_imoveis)
print("\n--- Imobiliária ---")
print(df_imoveis)

tipoCasa = df_imoveis['Tipo'] == 'Casa'
print(df_imoveis[tipoCasa])
print("\n")

valorAluguel = df_imoveis['Preco_Aluguel'] < 2000
bairroAluguel = df_imoveis['Bairro'] == 'Centro'

aluguelAcessivel = df_imoveis[bairroAluguel & valorAluguel]

print(aluguelAcessivel)
print("\n")

tamanho = df_imoveis['Area_m2'] > 130
casa = df_imoveis['Tipo'] == 'Casa'
praia = df_imoveis['Bairro'] == 'Praia'

cliente_praia = df_imoveis[praia & casa & tamanho]

print(cliente_praia)
print("\n")

tamanho = df_imoveis['Quartos'] == 1
tamanho2 = df_imoveis['Quartos'] == 2

sla = df_imoveis[tamanho | tamanho2]
print(sla)
print("\n")



```
